```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
model = LinearRegression()

LinearRegressionModel = model.fit(df[["Distance"]], df[["Time"]])
```

```python
LinearRegressionModel.predict(pd.DataFrame({"Distance": [55, 10]}))
```

```python
plt.scatter(df["Distance"], df["Time"], color='blue')
plt.xlabel('Distance')
plt.ylabel('Time')
plt.title('Distance vs Time')
plt.show()
```

```python
plt.scatter(df["Distance"], df["Time"], color='blue')
plt.plot(df["Distance"], LinearRegressionModel.predict(df[["Distance"]]), color='red')
plt.xlabel('Distance')
plt.ylabel('Time')
plt.title('Distance vs Time')
plt.show()
```

```python
# Coefficient (m)
# y = mx + b

print("Coefficient:", LinearRegressionModel.coef_)
```

```python
# Intercept (b)
# y = mx + b

print("Intercept:", LinearRegressionModel.intercept_)   
```
